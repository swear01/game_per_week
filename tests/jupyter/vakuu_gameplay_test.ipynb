# VakuuPlayer gameplay test

This notebook runs the recovered STS2 integration harness against the normal local mod environment. It verifies Preserved Fog, the ten relics, player control after auto-play, and auto-play on more than one turn.

The temporary `VakuuHarness` is deployed only for the test and is removed by the runner. During the run, only the local `VakuuPlayer` and `VakuuHarness` are enabled; the original mod settings and account profile are restored afterward.

```python
from pathlib import Path
import sys

candidates = [Path.cwd(), *Path.cwd().parents]
REPO_ROOT = next(path for path in candidates if (path / 'sls_vakuu/src/VakuuPlayer').exists())
ROOT = REPO_ROOT / 'sls_vakuu'
JUPYTER_DIR = REPO_ROOT / 'tests/jupyter'
sys.path.insert(0, str(JUPYTER_DIR))

from vakuu_test import GAME_APP, LOG_PATH, MOD_ROOT, game_pids, run_test

assert GAME_APP.exists(), GAME_APP
assert not game_pids(), 'Stop Slay the Spire 2 before starting the test'
print(f'root={ROOT}')
print(f'game={GAME_APP}')
print(f'log={LOG_PATH}')
print(f'mods={MOD_ROOT}')
```

```python
result = run_test()
assert len(result['auto_play_turns']) >= 2
print(result)
```
